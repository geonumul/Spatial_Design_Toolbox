## 1. 워크 문장으로 Word2Vec 학습  (따라 치기)

문자열 노드로 된 워크 세 개를 문장처럼 Word2Vec 에 넣어 노드마다 128칸짜리 숫자 벡터(임베딩)를 배워요.

- Word2Vec 은 "같은 문장 안에서 가까이 나오는 단어는 뜻이 비슷하다" 는 생각으로 단어마다 숫자 벡터를 배워요. node2vec 은 워크를 문장, 노드를 단어로 넣어요 (w1-walk 08).
- `vector_size=128` 은 벡터 칸 수, `window=5` 는 앞뒤 5개까지를 가까운 단어로 보기, `min_count=1` 은 한 번만 나온 단어도 쓰기, `sg=1` 은 skip-gram 방식, `epochs=100` 은 100번 반복이에요.
- 이름을 붙여 값을 넘기는 방식(`sentences=walks`)이에요 (c04). 괄호가 닫힐 때까지 한 문장이에요.
- gensim 은 사이트 브라우저에서 못 돌려서 Colab 에서 열려요. 노트북 앞에 설치 칸이 붙어요.

아래 코드를 **보면서 직접 쳐 보세요** (복사하지 말고요):

```python
from gensim.models import Word2Vec

walks = [["0", "1", "2", "1"], ["2", "3", "2", "0"], ["1", "0", "2", "3"]]

node2vec_model = Word2Vec(
    sentences=walks,
    vector_size=128,
    window=5,
    min_count=1,
    sg=1,
    epochs=100
)
vec0 = node2vec_model.wv["0"]
```

2. 입력 모양을 봐요. `walks` 는 리스트 안 리스트이고, 안쪽 칸은 숫자가 아니라 문자열 "0", "1" 이에요.
3. 순서대로 읽어요. gensim 에서 Word2Vec 을 불러오고, 설정값 여섯 개를 이름 붙여 넘겨 학습하고, `wv["0"]` 로 벡터를 꺼내요.

5. 확인: `print(vec0.shape)` 가 `(128,)` 이면 맞아요.

## 2. 임베딩을 표로 모으고 t-SNE 로 2차원  (빈칸 채우기)

노드마다 배운 128칸 벡터를 34 x 128 표로 모으고, t-SNE 로 34 x 2 표로 줄여 그림 그릴 준비를 해요.

- 원본 셀 10 은 PyG 의 `KarateClub()` 을 `to_networkx` 로 바꿔 써요. 여기서는 torch_geometric 설치 없이 networkx 에 들어 있는 `nx.karate_club_graph()` 의 엣지만 옮겨 같은 34명 그래프를 만들었어요(내려받는 것 없음). 반(class)은 PyG 의 4개 무리 대신 실제로 갈라진 두 동아리(Mr. Hi 쪽 0, 사범 쪽 1)를 써요.
- node2vec_weighted_choice, node2vec_walk 와 워크 만들기는 w1-walk 에서 짠 그대로 들어 있어요.
- `node2vec_model.wv[str(node)]` 는 노드 이름(문자열)으로 벡터를 꺼내요. 리스트 컴프리헨션으로 모아 `np.array` 로 감싸면 줄마다 노드 하나인 표가 돼요 (c03, c07).
- t-SNE 는 가까운 점은 가깝게 두면서 128차원 점들을 2차원 종이 위로 옮기는 방법이에요. `fit_transform` 이 옮긴 결과를 돌려줘요. `random_state=42` 로 결과를 고정해요.
- 원본 셀 13 은 이 뒤에 plt.scatter 로 점을 그리고 반마다 색을 칠해요. 여기서는 그리기 전까지만 해요.

<details><summary>생각 순서 보기 (힌트보다 먼저)</summary>

1. 입력은 학습된 `node2vec_model` 과 34명 그래프예요. 출력은 34 x 128 표 `node_embeddings`, 34 x 2 표 `node_embeddings_2d`, 반 번호 리스트 `labels` 예요.

5. 확인: `(34, 128) (34, 2)` 가 찍히면 맞아요.

````python
import random
import numpy as np
import networkx as nx
from gensim.models import Word2Vec
from sklearn.manifold import TSNE

club = nx.karate_club_graph()
karate_graph = nx.Graph()
karate_graph.add_nodes_from(range(34))
karate_graph.add_edges_from(club.edges())
for node in karate_graph.nodes():
    karate_graph.nodes[node]["class"] = 0 if club.nodes[node]["club"] == "Mr. Hi" else 1


def node2vec_weighted_choice(G, current_node, previous_node, p, q):
    neighbors = list(G.neighbors(current_node))
    unnormalized_weights = []

    for neighbor in neighbors:
        if neighbor == previous_node:
            unnormalized_weights.append(1.0 / p)
        elif G.has_edge(previous_node, neighbor):
            unnormalized_weights.append(1.0)
        else:
            unnormalized_weights.append(1.0 / q)

    norm_weights = [w / sum(unnormalized_weights) for w in unnormalized_weights]
    return random.choices(neighbors, weights=norm_weights)[0]

def node2vec_walk(G, start_node, num_steps, p, q):
    walk = [start_node]
    for _ in range(num_steps - 1):
        current_node = walk[-1]
        neighbors = list(G.neighbors(current_node))

        if len(neighbors) == 0:
            break

        if len(walk) == 1:
            next_node = random.choice(neighbors)
        else:
            next_node = node2vec_weighted_choice(G, current_node, walk[-2], p, q)

        walk.append(next_node)
    return walk


random.seed(0)
# Generate random walks
num_walks = 100
walk_length = 10
p = 0.8
q = 0.2

walks = [
    [str(node) for node in node2vec_walk(
        karate_graph,
        start_node=random.choice(list(karate_graph.nodes())),
        num_steps=walk_length,
        p=p,
        q=q
    )]
    for _ in range(num_walks)
]

# 3. Train Word2Vec Model
node2vec_model = Word2Vec(
    sentences=walks,
    vector_size=128,
    window=5,
    min_count=1,
    sg=1,
    epochs=100
)

# Extract embeddings
node_embeddings = np.array([node2vec_model.wv[___(node)] for node in karate_graph.nodes()])

# 4. Visualize with t-SNE
tsne = TSNE(n_components=___, random_state=42)
node_embeddings_2d = tsne.fit_transform(___)

labels = [karate_graph.nodes[node][___] for node in karate_graph.nodes()]
print(node_embeddings.shape, node_embeddings_2d.shape)
````

## 3. 음성 샘플에 출발, 도착 노드가 섞여요  (고치기)

엣지 0-1 의 음성 샘플을 뽑았는데 0 이나 1 이 섞여 나와요. 조건 한 곳을 고쳐요.

- 음성 샘플(Negative Sample)은 "이 엣지와 상관없는 노드" 여야 해요. 그래서 출발 노드(sourcenode)도 아니고 도착 노드(targetnode)도 아닐 때만 넣어요.
- "A 도 아니고 B 도 아니다" 는 `x != A and x != B` 예요. `or` 로 쓰면 둘 중 하나만 달라도 참이라서 거의 모든 노드가 들어가요 (c03 의 and, or).
- `while len(negsamples) < negsamplesize:` 는 원하는 개수가 찰 때까지 계속 뽑아요. `torch.tensor(..., dtype=torch.long)` 은 노드 번호 텐서예요 (c09).
- VoseAlias 는 04 에서 완성한 그대로 들어 있어요.

<details><summary>생각 순서 보기 (힌트보다 먼저)</summary>

1. 증상부터 봐요. 에러는 없는데 엣지 0-1 의 음성 샘플에 0 이나 1 이 섞여 나와요.
2. 샘플을 넣는 곳은 `if` 한 줄뿐이에요. 조건을 손으로 풀어요. 뽑힌 노드가 0 이면 `0 != 0` 은 거짓, `0 != 1` 은 참이에요.
3. `or` 는 둘 중 하나만 참이어도 참이라서 0 이 들어가요. 두 노드 모두와 달라야 하니 `and` 로 바꿔요.
4. 고칠 곳은 조건 한 줄이에요. VoseAlias 는 04 에서 확인한 그대로라 건드리지 않아요.
5. 확인: 출력 텐서 5칸이 모두 2 나 3 이면 맞아요.

    뽑기표로 노드 하나 samplednode 를 뽑는다
    samplednode 가 sourcenode 와도 다르고 targetnode 와도 다르면
        negsamples 에 넣는다

## 5. LINE 모델 클래스 완성  (빈칸 채우기)

이어진 두 노드 벡터의 곱은 크게, 음성 샘플과의 곱은 작게 만드는 손실을 계산하는 LINE 클래스를 완성해요.

- 손실 식은 $-\left[\log\sigma(v_i \cdot v_j) + \sum_{k} \log\sigma(-v_i \cdot v_k)\right]$ 의 평균이에요. $\sigma$ 는 0~1 로 눌러 주는 시그모이드, $v_k$ 는 음성 샘플 벡터예요.
- order 1(1차 근접성)은 이웃 벡터도 같은 표 `nodes_embeddings` 에서, order 2(2차 근접성)는 "문맥" 표 `contextnodes_embeddings` 에서 꺼내요.
- `torch.mul(a, b)` 는 칸끼리 곱, `torch.sum(..., dim=1)` 은 줄마다 더하기라서 둘을 합치면 줄마다 내적이에요 (c09).
- `v_i.view(len(v_i), 1, self.embed_dim)` 은 (B, D) 를 (B, 1, D) 로 바꿔서 음성 샘플 (K, D) 와 곱하면 (B, K, D) 가 되게 해요. 그래서 dim=2 로 더해요.
- `super(LINE, self).__init__()` 는 `super().__init__()` 의 옛날 쓰는 법이에요. 뜻은 같아요 (c06).

<details><summary>생각 순서 보기 (힌트보다 먼저)</summary>

1. 입력은 엣지 앞쪽 번호 `v_i` (B,), 뒤쪽 번호 `v_j` (B,), 음성 샘플 번호 `negsamples` (K,) 예요. 출력은 손실 숫자 하나예요.
2. 모양을 적어 가며 따라가요. 번호로 표에서 꺼내면 v_i, v_j 는 (B, D), negativenodes 는 (K, D) 가 돼요.
3. 양성 쪽: 칸끼리 곱해 줄마다(dim=1) 더하면 엣지마다 내적 하나, (B,) 예요.
4. 음성 쪽: v_i 를 (B, 1, D) 로 바꿔 곱하면 (B, K, D), dim=2 로 더하면 (B, K), logsigmoid 뒤 dim=1 로 더하면 (B,) 예요.
5. 틀리기 쉬운 곳: order 2 일 때 v_j 는 문맥 표 `contextnodes_embeddings` 에서 꺼내요. 손실은 작을수록 좋은 값이라 평균에 마이너스를 붙여요.
6. 확인: B=2, K=3, D=8 테스트에서 숫자 하나가 약 2.77 로 찍히면 맞아요. 처음 벡터가 0 에 가까워서 logsigmoid 가 약 -0.69 씩 4번 더해진 값이에요.

## 6. 내적 합의 방향이 반대예요  (고치기)

엣지 3개를 한꺼번에 넣으면 모양 에러가 나요(엣지 1개일 때는 에러 없이 틀린 값). 양성 쪽 torch.sum 의 dim 을 고쳐요.

- v_i, v_j 는 (엣지 수 B, 칸 수 D) 표예요. 엣지마다 내적 하나를 원하니 **줄마다**, 즉 dim=1 로 더해야 (B,) 가 나와요.
- dim=0 으로 더하면 칸마다 더해져 (D,) 가 돼요. 음성 쪽 결과 (B,) 와 더할 때 모양이 안 맞아요 (c09, c11-03 의 dim 반대 실수).
- B 가 1 이면 (D,) + (1,) 이 에러 없이 더해져서 더 찾기 어려워요.

<details><summary>생각 순서 보기 (힌트보다 먼저)</summary>

1. 에러 메시지 마지막 줄부터 봐요. `The size of tensor a (8) must match the size of tensor b (3)` 라고 모양이 안 맞는다고 해요.
2. 어느 줄인지 찾아요. 줄 번호가 `loss = positivebatch + negativebatch` 를 가리켜요.
3. 그 줄 앞에 `print(positivebatch.shape, negativebatch.shape)` 를 넣으면 `torch.Size([8]) torch.Size([3])` 이 나와요. 엣지가 3개니 둘 다 (3,) 이어야 해요.
4. 8 은 벡터 칸 수 D 예요. 줄(엣지)마다 더할 것을 칸마다 더했어요. 양성 쪽 `dim=0` 을 `dim=1` 로 바꿔요.
5. 확인: 에러 없이 손실 숫자 하나가 찍히면 맞아요.

## 7. train_line 직접 짜기  (직접 짜기)

주석만 보고 LINE 학습 함수 train_line 을 짜요. 엣지마다 음성 샘플을 뽑고, 손실을 줄이도록 한 걸음씩 고쳐요.

- 원본 셀 10 은 PyG 의 `KarateClub()` 을 `to_networkx` 로 바꿔 써요. 여기서는 torch_geometric 설치 없이 networkx 에 들어 있는 `nx.karate_club_graph()` 의 엣지만 옮겨 같은 34명 그래프를 만들었어요(내려받는 것 없음). 반(class)은 PyG 의 4개 무리 대신 실제로 갈라진 두 동아리(Mr. Hi 쪽 0, 사범 쪽 1)를 써요.
- 원본은 epochs=100 이지만 채점 시간을 줄이려고 10 으로 했어요. 함수 모양은 같아요.
- c11 학습 루프와 같은 네 줄(zero_grad, 손실, backward, step)을 **엣지 하나마다** 해요. 그래서 반복이 두 겹이에요: 바깥 epoch, 안쪽 edges.
- `torch.device('cuda' if torch.cuda.is_available() else 'cpu')` 는 GPU 가 있으면 GPU 를 써요. `.to(device)` 로 모델과 텐서를 같은 곳에 둬요.
- 채점은 seed 를 고정하고 두 그래프(가라테 order 1, 6각형 order 2)에서 학습 결과 표가 원본과 같은지 봐요. 무작위를 쓰는 순서가 원본과 같아야 해요: 모델 만들기 → 옵티마이저 → makeDist → VoseAlias → 반복.

1. 입력은 그래프와 설정값(embed_dim, epochs, neg_samples, order), 출력은 학습이 끝난 LINE 모델이에요.

## 8. LINE 임베딩을 2차원으로 꺼내기  (직접 짜기)

학습한 LINE 모델에서 노드 벡터 표를 numpy 로 꺼내 t-SNE 로 2차원으로 줄이고, 노드마다 반 번호 리스트와 함께 돌려주는 함수를 짜요.

- 원본 셀 18 visualize_line_embeddings 에서 그림 그리기 앞부분을 함수로 만든 거예요. 그림 코드(plt.scatter 등) 대신 두 결과를 return 해요.
- `model.nodes_embeddings.weight` 는 34 x 128 표예요. `.cpu().detach().numpy()` 로 GPU 에서 꺼내고, 계산 기록을 떼고, numpy 로 바꿔요 (c09).
- t-SNE 설정은 02 와 같아요: `TSNE(n_components=2, random_state=42)` 와 `fit_transform`.
- 채점은 서로 다른 두 모델(order 1, order 2)로 원본 방식 결과와 비교해요. 학습 시간을 줄이려고 학습하지 않은 모델을 넣어요.

<details><summary>생각 순서 보기 (힌트보다 먼저)</summary>

1. 입력은 LINE 모델과 그래프, 출력은 (34, 2) numpy 표 `embeddings_2d` 와 노드별 반 번호 리스트 `labels` 두 개예요.
2. 모양으로 따라가요. `model.nodes_embeddings.weight` 는 (34, 128) 텐서예요. numpy 로 바꾼 뒤 t-SNE 를 거치면 (34, 2) 가 돼요.
3. 텐서를 numpy 로 바꾸는 길은 세 걸음이에요. `.cpu()` 로 GPU 에서 내리고, `.detach()` 로 학습 기록을 떼고, `.numpy()` 로 바꿔요.
4. 틀리기 쉬운 곳: `.detach()` 를 빼면 `Can't call numpy() on Tensor that requires grad` 에러가 나요. 가중치에는 학습 기록이 붙어 있어요.
5. 확인: `(34, 2) [0, 0, 0, 0, 0]` 이 찍히면 맞아요.

````python
# 채점 칸: 위 칸을 실행한 다음 이 칸을 실행하세요
import collections
import random
from decimal import Decimal
import numpy as np
import networkx as nx
import torch
import torch.nn as nn
import torch.nn.functional as F

class _RVoseAlias:
    def __init__(self, dist):
        self.dist = dist
        self.alias_initialisation()

    def alias_initialisation(self):
        n = len(self.dist)
        self.table_prob = {}
        self.table_alias = {}
        scaled_prob = {}
        small = []
        large = []

        for o, p in self.dist.items():
            scaled_prob[o] = Decimal(p) * n
            if scaled_prob[o] < 1:
                small.append(o)
            else:
                large.append(o)

        while small and large:
            s = small.pop()
            l = large.pop()

            self.table_prob[s] = scaled_prob[s]
            self.table_alias[s] = l
            scaled_prob[l] = (scaled_prob[l] + scaled_prob[s]) - Decimal(1)

            if scaled_prob[l] < 1:
                small.append(l)
            else:
                large.append(l)

        while large:
            self.table_prob[large.pop()] = Decimal(1)
        while small:
            self.table_prob[small.pop()] = Decimal(1)

        self.listprobs = list(self.table_prob)

    def alias_generation(self):
        col = random.choice(self.listprobs)
        if self.table_prob[col] >= random.uniform(0, 1):
            return col
        else:
            return self.table_alias[col]

    def sample_n(self, size):
        for i in range(size):
            yield self.alias_generation()

def _RmakeDist(graph: nx.Graph, power=0.75):
    edgedistdict = collections.defaultdict(int)
    nodedistdict = collections.defaultdict(int)

    weightsum = 0
    negprobsum = 0

    for edge in graph.edges(data=True):
        node1, node2, weight = edge[0], edge[1], edge[2].get("weight", 1)
        edgedistdict[tuple([node1, node2])] = weight
        nodedistdict[node1] += weight
        weightsum += weight
        negprobsum += np.power(weight, power)

    for node, outdegree in nodedistdict.items():
        nodedistdict[node] = np.power(outdegree, power) / negprobsum

    for edge, weight in edgedistdict.items():
        edgedistdict[edge] = weight / weightsum

    return nodedistdict

def _RnegSampleBatch(sourcenode, targetnode, negsamplesize, nodesaliassampler):
    negsamples = []
    while len(negsamples) < negsamplesize:
        samplednode = nodesaliassampler.alias_generation()
        if samplednode != sourcenode and samplednode != targetnode:
            negsamples.append(samplednode)
    return torch.tensor(negsamples, dtype=torch.long)

class _RLINE(nn.Module):
    def __init__(self, size, embed_dim=128, order=1):
        super(_RLINE, self).__init__()
        assert order in [1, 2], "Order should either be 1 or 2"
        self.embed_dim = embed_dim
        self.order = order

        self.nodes_embeddings = nn.Embedding(size, embed_dim)
        self.nodes_embeddings.weight.data = self.nodes_embeddings.weight.data.uniform_(-.5, .5) / embed_dim

        if order == 2:
            self.contextnodes_embeddings = nn.Embedding(size, embed_dim)
            self.contextnodes_embeddings.weight.data = self.contextnodes_embeddings.weight.data.uniform_(-.5, .5) / embed_dim

    def forward(self, v_i, v_j, negsamples, device):
        v_i = self.nodes_embeddings(v_i).to(device)

        if self.order == 2:
            v_j = self.contextnodes_embeddings(v_j).to(device)
            negativenodes = -self.contextnodes_embeddings(negsamples).to(device)
        else:
            v_j = self.nodes_embeddings(v_j).to(device)
            negativenodes = -self.nodes_embeddings(negsamples).to(device)

        mulpositivebatch = torch.mul(v_i, v_j)
        positivebatch = F.logsigmoid(torch.sum(mulpositivebatch, dim=1))

        mulnegativebatch = torch.mul(v_i.view(len(v_i), 1, self.embed_dim), negativenodes)
        negativebatch = torch.sum(F.logsigmoid(torch.sum(mulnegativebatch, dim=2)), dim=1)

        loss = positivebatch + negativebatch
        return -torch.mean(loss)

def _Rtrain_line(graph, embed_dim=128, epochs=10, neg_samples=5, order=1):
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    model = _RLINE(size=len(graph.nodes), embed_dim=embed_dim, order=order).to(device)
    optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

    nodedistdict = _RmakeDist(graph)
    nodesaliassampler = _RVoseAlias(nodedistdict)

    edges = list(graph.edges())

    model.train()
    for epoch in range(epochs):
        total_loss = 0
        for v_i, v_j in edges:
            v_i_tensor = torch.tensor([v_i], dtype=torch.long).to(device)
            v_j_tensor = torch.tensor([v_j], dtype=torch.long).to(device)
            neg_samples_batch = _RnegSampleBatch(v_i, v_j, neg_samples, nodesaliassampler).to(device)

            optimizer.zero_grad()
            loss = model(v_i_tensor, v_j_tensor, neg_samples_batch, device)
            loss.backward()
            optimizer.step()
            total_loss += loss.item()


    return model

_club = nx.karate_club_graph()
_KG = nx.Graph()
_KG.add_nodes_from(range(34))
_KG.add_edges_from(_club.edges())
for _nd in _KG.nodes():
    _KG.nodes[_nd]["class"] = 0 if _club.nodes[_nd]["club"] == "Mr. Hi" else 1
import numpy as np
from sklearn.manifold import TSNE
assert 'line_embeddings_2d' in globals(), "line_embeddings_2d 함수가 있어야 해요"
for _seed, _order, _dim in [(2, 1, 16), (5, 2, 32)]:
    torch.manual_seed(_seed)
    _m = _RLINE(34, embed_dim=_dim, order=_order)
    try:
        _res = line_embeddings_2d(_m, _KG)
    except Exception as _e:
        raise AssertionError(f"line_embeddings_2d 를 부르다가 에러가 났어요: {_e}")
    assert isinstance(_res, tuple) and len(_res) == 2, "embeddings_2d, labels 두 값을 return 해야 해요"
    _e2, _lb = _res
    assert isinstance(_e2, np.ndarray) and _e2.shape == (34, 2), f"embeddings_2d 는 34 x 2 numpy 표여야 해요. 지금은 {getattr(_e2, 'shape', None)}"
    _ref = TSNE(n_components=2, random_state=42).fit_transform(_m.nodes_embeddings.weight.cpu().detach().numpy())
    assert np.allclose(_e2, _ref, atol=1e-3), "t-SNE 결과가 원본과 달라요. nodes_embeddings.weight 를 넣었는지, random_state=42 인지 확인해요"
    assert list(_lb) == [_KG.nodes[_n]["class"] for _n in _KG.nodes], "labels 는 graph.nodes[node]['class'] 리스트예요"
    assert not _m.training, "model.eval() 로 평가 모드로 바꿔요"

print("통과! 과제 1 Part 2 를 끝까지 직접 쓸 수 있게 됐어요. 마지막 plt.scatter 에 이 두 값을 넣으면 셀 18 그림이 나와요.")
````
