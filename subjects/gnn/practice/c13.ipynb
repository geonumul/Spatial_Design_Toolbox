## 1. 같은 뼈대를 torch 로 (Colab)  (직접 짜기)

- 순수 파이썬 train_w 의 줄마다 짝이 있어요. 기울기 식은 `loss.backward()` 가 대신 계산해요.
- `.item()` 은 숫자 하나짜리 텐서를 파이썬 숫자로 꺼내요.

<details><summary>정답 보기</summary>

```python
import torch
import torch.nn as nn

torch.manual_seed(0)
x = torch.tensor([[1.0], [2.0], [3.0]])
y = torch.tensor([[2.0], [4.0], [6.0]])
model = nn.Linear(1, 1, bias=False)
opt = torch.optim.SGD(model.parameters(), lr=0.05)
for epoch in range(200):
    opt.zero_grad()
    pred = model(x)
    loss = ((pred - y) ** 2).mean()
    loss.backward()
    opt.step()

w_final = model.weight.item()
print(w_final)
```
