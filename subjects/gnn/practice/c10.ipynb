## 1. PyTorch 가 준 기본 틀 물려받기  (따라 치기)

wc-6 에서 직접 만든 Layer 부모 틀 대신, PyTorch 가 준 기본 틀 nn.Module 을 물려받아 "받은 값에 2 곱하기" 모델을 만들어요.

- 이 세트는 Google Colab 에서 열려요. 코드 칸에 치고 실행(Shift+Enter)한 뒤, 아래 채점 칸을 실행하면 돼요.
- `import torch.nn as nn` 은 PyTorch 의 신경망 도구 상자(torch.nn)를 `nn` 이라는 짧은 이름으로 불러와요.
- `class Double(nn.Module):` 는 **PyTorch 가 준 기본 붕어빵 틀 nn.Module** 을 물려받는다는 뜻이에요. wc-6 의 `class MyLayer(Layer):` 와 같은 모양이에요.
- `super().__init__()` 는 "부모 틀(nn.Module)의 준비 작업을 먼저 해 줘" 라는 줄이에요. nn.Module 을 물려받으면 `__init__` 첫 줄에 꼭 써요.
- `forward` 는 데이터가 들어오면 할 일이에요. 여기서는 받은 값 x 에 2를 곱해 돌려줘요.
- `model(텐서)` 처럼 객체를 함수처럼 부르면, wc-6 의 `__call__` 처럼 nn.Module 이 알아서 forward 를 불러 줘요.

아래 코드를 **보면서 직접 쳐 보세요** (복사하지 말고요):

```python
import torch
import torch.nn as nn

class Double(nn.Module):
    def __init__(self):
        super().__init__()

    def forward(self, x):
        return x * 2

model = Double()
out = model(torch.tensor([1.0, 2.0, 3.0]))
```

## 3. 층을 속성으로 달기: nn.Linear  (따라 치기)

숫자 3개를 받아 2개로 바꾸는 층(nn.Linear)을 모델 안에 속성으로 달고, 노드 4개짜리 입력을 통과시켜요.

- `nn.Linear(3, 2)` 는 **숫자 3개를 받아 숫자 2개로 바꾸는 층**이에요. 속에 곱할 숫자 표(가중치, weight)와 더할 숫자(bias)가 들어 있고, 학습하면서 이 숫자들이 바뀌어요.
- `self.lin = nn.Linear(3, 2)` 는 붕어빵에 속을 넣듯이, 모델 객체에 층을 `lin` 이라는 이름으로 달아 두는 줄이에요.
- `forward` 안의 `self.lin(x)` 는 달아 둔 층에 x 를 통과시켜요. 층도 nn.Module 이라 함수처럼 부를 수 있어요.
- `torch.ones(4, 3)` 은 1로 채운 4행 3열 표예요. 행 하나가 노드(데이터) 하나, 열이 특징 하나라고 생각해요. 결과 out 은 4행 2열이 돼요.
- `torch.manual_seed(0)` 은 층의 무작위 시작 숫자를 매번 똑같이 만드는 줄이에요(wc-4 의 random.seed 와 같은 뜻).

아래 코드를 **보면서 직접 쳐 보세요** (복사하지 말고요):

```python
import torch
import torch.nn as nn

class OneLayer(nn.Module):
    def __init__(self):
        super().__init__()
        self.lin = nn.Linear(3, 2)
    def forward(self, x):
        return self.lin(x)

torch.manual_seed(0)
model = OneLayer()
out = model(torch.ones(4, 3))
```

## 5. 고치기: super().__init__() 가 빠졌어요  (고치기)

실행하면 AttributeError 가 나요. 에러 메시지를 읽고 빠진 한 줄을 넣어 고쳐요.

- 에러가 나는 건 정상이에요. 메시지 **마지막 줄**부터 읽어요: `cannot assign module before Module.__init__() call` 는 "부모 준비 작업(Module.__init__)을 부르기 전에 층을 달 수 없어요" 라는 뜻이에요.
- nn.Module 은 `super().__init__()` 에서 층을 넣어 둘 보관함을 만들어요. 이 줄이 없으면 `self.lin = ...` 을 넣을 곳이 없어요.
- 그래서 `super().__init__()` 는 반드시 `self.lin = ...` **보다 위**, `__init__` 첫 줄에 써요.

<details><summary>힌트 1</summary>

에러 메시지에 나오는 Module.__init__() 를 부르는 방법이 무엇이었죠?

</details>

<details><summary>힌트 2</summary>

__init__ 안, self.lin 줄 위에 한 줄을 넣어요

</details>

<details><summary>힌트 3</summary>

super().__init__()

</details>


````python
import torch
import torch.nn as nn

class Tiny(nn.Module):
    def __init__(self):
        self.lin = nn.Linear(3, 1)

    def forward(self, x):
        return self.lin(x)

torch.manual_seed(0)
model = Tiny()
out = model(torch.ones(2, 3))
print(out)
````

````python
# 채점 칸: 위 칸을 실행한 다음 이 칸을 실행하세요
import torch
import torch.nn as nn
assert 'model' in globals(), "model = Tiny() 까지 실행되어야 해요. super().__init__() 를 __init__ 첫 줄에 넣어 보세요"
assert isinstance(getattr(model, 'lin', None), nn.Linear), "self.lin = nn.Linear(3, 1) 은 그대로 두세요"
assert len(list(model.parameters())) == 2, "층이 모델에 제대로 등록되어야 해요. super().__init__() 가 self.lin 보다 위에 있어야 해요"
assert 'out' in globals() and tuple(out.shape) == (2, 1), "out 모양은 (2, 1) 이어야 해요"
print("통과! 초보가 가장 많이 하는 실수예요. 2주차 노트북 모든 클래스의 `__init__` 첫 줄이 `super().__init__()` 인 이유예요.")
````

## 6. 고치기: model.forward(x) 말고 model(x)  (고치기)

모델을 `model.forward(x)` 로 직접 불렀어요. 결과 숫자는 같아 보여도 PyTorch 가 해 주는 준비 과정을 건너뛰어요. `model(x)` 로 고쳐요.

- wc-6 에서 `__call__` 이 forward 를 부르게 만들었죠. nn.Module 의 `__call__` 은 forward 를 부르기 **전후에** PyTorch 의 부가 작업(기록 장치 등)도 함께 해요.
- 그래서 약속은 항상 `model(x)` 예요. `model.forward(x)` 로 부르면 그 부가 작업이 빠져요.
- 시작 코드의 `calls` 두 줄은 채점용 기록 장치예요. `model(x)` 로 부를 때만 calls 에 1 이 쌓여요. 이해하지 않아도 되니 지우지만 마세요.

<details><summary>힌트 1</summary>

forward 를 직접 부르는 줄을 찾아요

</details>

<details><summary>힌트 2</summary>

객체를 함수처럼 부르면 __call__ 이 forward 를 대신 불러요

</details>

<details><summary>힌트 3</summary>

out = model(x)

</details>


````python
import torch
import torch.nn as nn

class Small(nn.Module):
    def __init__(self):
        super().__init__()
        self.lin = nn.Linear(2, 2)

    def forward(self, x):
        return self.lin(x)

torch.manual_seed(0)
model = Small()
# 채점용 기록 장치 (지우지 마세요): model(x) 로 부를 때만 calls 에 1 이 쌓여요
calls = []
model.register_forward_hook(lambda m, inp, outp: calls.append(1))

x = torch.ones(3, 2)
out = model.forward(x)
print(out.shape, len(calls))
````

````python
# 채점 칸: 위 칸을 실행한 다음 이 칸을 실행하세요
import torch
assert 'calls' in globals(), "calls = [] 기록 장치 줄은 지우지 마세요"
assert 'out' in globals() and torch.is_tensor(out), "out = model(x) 가 있어야 해요"
assert tuple(out.shape) == (3, 2), f"out 모양은 (3, 2) 여야 해요. 지금은 {tuple(out.shape)} 예요"
assert len(calls) == 1, f"model(x) 로 한 번 불러야 기록 장치가 한 번 켜져요. 지금 기록은 {len(calls)}번이에요. model.forward(x) 를 model(x) 로 바꿔 보세요"
print("통과! 2주차 노트북도 전부 `model(x, support)`, `self.conv1(x, edge_index)` 처럼 부르고 `.forward(` 는 쓰지 않아요.")
````

## 9. ReLU: F.relu 와 torch.relu  (따라 치기)

import torch
